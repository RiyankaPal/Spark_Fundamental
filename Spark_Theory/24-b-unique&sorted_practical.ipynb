```python
data=[(10 ,'Anil',50000, 18),
(11 ,'Vikas',75000,  16),
(12 ,'Nisha',40000,  18),
(13 ,'Nidhi',60000,  17),
(14 ,'Priya',80000,  18),
(15 ,'Mohit',45000,  18),
(16 ,'Rajesh',90000, 10),
(17 ,'Raman',55000, 16),
(18 ,'Sam',65000,   17),
(15 ,'Mohit',45000,  18),
(13 ,'Nidhi',60000,  17),      
(14 ,'Priya',90000,  18),  
(18 ,'Sam',65000,   17)
]
schema = ['id','name','sal','mngr_id']
manager_df = spark.createDataFrame(data=data,schema=schema)



```

```python
manager_df.show()
```

```python
manager_df.distinct().show()
```

```python
manager_df.select("id","name").distinct().show()
```

```python
manager_df.dropDuplicates(["id","name"]).show()
```

```python
dropped_mngr_data = manager_df.dropDuplicates(["id","name","sal","mngr_id"]).show()
```

```python
from pyspark.sql.functions import *
```

```python
manager_df.sort(col("sal")).show()
```

```python
manager_df.sort(col("sal").desc()).show()
```

```python
manager_df.sort(col("sal").desc(),col("name").desc()).show()
```
