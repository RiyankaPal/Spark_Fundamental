## How to read line delimited json file?

Name_data.json:
```python
{"name":"Manish","age":20,"salary":20000},
{"name":"Nikita","age":25,"salary":21000},
{"name":"Pritam","age":16,"salary":22000},
{"name":"Prantosh","age":35,"salary":25000},
{"name":"Vikash","age":67,"salary":40000}
```




````python
spark.read.format("json")\
    .option("inferschema","True")\
    .option("mode","PERMISSIVE")\
    .load("/FileStore/tables/name_data.json").show()
````

## What if we have 3 keys in all line but 4 key in one line?
single_file_json_with_extra_fields:
```python
{"name":"Manish","age":20,"salary":20000},
{"name":"Nikita","age":25,"salary":21000},
{"name":"Pritam","age":16,"salary":22000},
{"name":"Prantosh","age":35,"salary":25000},
{"name":"Vikash","age":67,"salary":40000,"gender":"M"}
```

In above data  there is an extra field in last line. Now we will check how spark read this type of data:

````python
spark.read.format("json")\
    .option("inferschema","True")\
    .option("mode","PERMISSIVE")\
    .load("/FileStore/tables/single_file_json_with_extra_fields.json").show()
````

## How to read multi line json file?
Multi_line_correct:
```python
[
{
  "name": "Manish",
  "age": 20,
  "salary": 20000
},
{
  "name": "Nikita",
  "age": 25,
  "salary": 21000
},
{
  "name": "Pritam",
  "age": 16,
  "salary": 22000
},
{
  "name": "Prantosh",
  "age": 35,
  "salary": 25000
},
{
  "name": "Vikash",
  "age": 67,
  "salary": 40000
}
]
```
To read multi line json file we have to set the multiline option *True*

````python
spark.read.format("json")\
    .option("inferschema","True")\
    .option("mode","PERMISSIVE")\
    .option("multiline","True")\
    .load("/FileStore/tables/Multi_line_correct.json").show()
````

## Which one works faster- line delimited or multiline json file?
Ans: Line-delimited JSON works faster. In the case of multi-line JSON, Spark treats the entire file as a single object, which increases the overhead during read and write operations. In contrast, with line-delimited JSON, Spark processes the data line by line, making parsing and processing significantly more efficient.
