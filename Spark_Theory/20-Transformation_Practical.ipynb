```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType
```

```python
from pyspark.sql.functions import col,expr,lit
```

##### How To Create Schema

```python

employee_schema = StructType(fields=[StructField("Id",IntegerType(),False),
                                     StructField("name",StringType(),True),
                                     StructField("age",IntegerType(),True),
                                     StructField("salary",IntegerType(),True),
                                     StructField("address",StringType(),True),
                                     StructField("gender",StringType(),True)
])
```

#### How to read files

```python

Employee_df=spark.read\
    .option("header",True)\
    .schema(employee_schema)\
    .csv(f"{raw_folder_path}/employee_write_data.csv")
```

```python
##### How Many Ways To Select Columns
```

```python
Employee_df.select("name").show()
```

```python
Employee_df.select(col("id") + 5).show()
```

```python
Employee_df.select("name","age","Id").show()
```

```python
Employee_df.select(col("name"),col("age"),col("Id")).show()
```

```python
Employee_df.select("id",col("name"),Employee_df["salary"],Employee_df.address).show()
```

## Expression

```python
Employee_df.select(expr("id + 5 ")).show()
```

```python
Employee_df.select(expr("id as employee_id"),expr("name as employee_name"),expr("concat(name,address)")).show()
```

```python
Employee_df.select("*").show()
```

Spark SQL

```python
Employee_df.createOrReplaceTempView("employee_tbl")
```

```python
spark.sql("""
          SELECT * 
          FROM employee_tbl
          """).show()
```

#### Alias

```python
Employee_df.select(col("id").alias("employee_id"),"name").show()
```

```python
Employee_df.filter(col("salary")>150000).show()
```

```python
Employee_df.where(col("salary")>150000).show()
```

```python
Employee_df.filter((col("salary")>150000) & (col("age")<18)).show()
```

```python
Employee_df.select("*",lit("Kumar").alias("Last_Name")).show()
```

##### How To Add columns

```python
Employee_df.withColumn("Sur_name",lit("singh")).show()
```

##### How to Rename Column

```python
Employee_df.withColumnRenamed("id","Employee_Id").show()
```

##### How to cast

```python
Employee_df.withColumn("id",col("id").cast("string"))\
           .withColumn("salary",col("salary").cast("long"))\
             .printSchema()
```

```python
Employee_df.drop("id",col("name")).show()
```

```python
spark.sql("""
          Select *,"kumar" as last_name, concat(name,last_name) as full_name, cast(id as string) 
          From employee_tbl
          Where salary >150000 AND age < 18
          """).show()
```
