```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType
```

```python
from pyspark.sql.functions import col,when,lit
```

```python
employee_schema = StructType(fields=[StructField("Id",IntegerType(),False),
                                     StructField("name",StringType(),True),
                                     StructField("age",IntegerType(),True),
                                     StructField("salary",IntegerType(),True),
                                     StructField("address",StringType(),True),
                                     StructField("gender",StringType(),True)
])
```

```python
Employee_df=spark.read\
    .option("header",True)\
    .schema(employee_schema)\
    .csv(f"{raw_folder_path}/employee_write_data.csv")
```

```python
Employee_df.show()
```

```python
Employee_df.withColumn("adult",when(col("age")<18,"No")
                               .when(col("age")>18,"Yes")
                               .otherwise("Novalue")).show()
```

```python
Employee_df.withColumn("age",when(col("age").isNull(),lit(19))
                  .otherwise(col("age")))\
      .withColumn("adult",when(col("age")>18,"Yes")
                  .otherwise("No")).show()
```

```python
Employee_df.withColumn("age_wise",when((col("age")>0) & (col("age")<18),"Minor")
                       .when((col("age")>18) & (col("age")<30),"Mid")
                       .otherwise("major")).show()
```
