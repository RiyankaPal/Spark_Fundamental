Schema: We can crate schema in pyspark in 2 way:
- **StructType & StructField** -> StructType : Which defines our structure of DataFrame.This contains list of structfield AND
StructField: structure of columns.<br>
- **DDL**-> Example: ddl_my_Schema="id integer,name string, age integer"

```python
from pyspark.sql.types import StructField,StructType,StringType,IntegerType # to import StructType & StructField.
```

- schema Syntax

```python
my_schema=StructType(
                         [
                             StructField("DEST_COUNTRY_NAME",StringType(),True),
                             StructField("ORIGIN_COUNTRY_NAME",StringType(),True),
                             StructField("COUNT",IntegerType(),True)


                         ]) # We are defining our schema through StructType & StructField.             
```

```python
FLIGHT_DF_schema=spark.read.format("csv")\
            .option("header", "false")\
            .option("inferschema","false")\
            .schema(my_schema)\
            .option("mode","PERMISSIVE")\
            .load("/FileStore/tables/2010_summary.csv")
FLIGHT_DF_schema.show(5)
```

1. As header is false so it is taking header as data and showing count as NULL because PERMISSIVE MODE<br>

2. To fix above issue we will use skiprows. <br>

```python
FLIGHT_DF_schema=spark.read.format("csv")\
            .option("header", "false")\
            .option("skipRows",1)\
            .option("inferschema","false")\
            .schema(my_schema)\
            .option("mode","PERMISSIVE")\
            .load("/FileStore/tables/2010_summary.csv")
FLIGHT_DF_schema.show(5)
```
