**format:** Data File Format like: json,csv,JDBC/ODBC,Table,parquet <br>
(format field is optional by default it support parquet)<br>
**Option:** like inferschema, mode, header (Option field is also optional)<br>
**schema:** We can pass the data types through schema. (optional field)<br>
**load:** Here we pass the data path, where our data is residing.(**not optional**)<br>

**MODE :**
1. **Failfast:** fail execution if malformed record in dataset.
2. **Dropmalformed:** Drop the corrupted record.
3. **Permissive:** this is default . This mode will set the null value to all corrupted fields.




````python
FLIGHT_DF=spark.read.format("csv")\
        .option("header", "false")\
        .option("inferschema","false")\
        .option("mode","FAILFAST")\
        .load("/FileStore/tables/2010_summary.csv")
FLIGHT_DF.show(5)
````

**LINE 2**: 
```python 
.option("header", "false") :
``` 
Here Header is false so, not showing the Header in OUTPUT.



````python
FLIGHT_DF_header=spark.read.format("csv")\
            .option("header", "true")\
            .option("inferschema","false")\
            .option("mode","FAILFAST")\
            .load("/FileStore/tables/2010_summary.csv")
FLIGHT_DF_header.show(5)
````

**LINE 2:** As Header is True we can see now Header.

````python
FLIGHT_DF_header.printSchema() # printschema shows the schema or metadata of columns 
                               #As inferschema is false it is showing count as string. means now it can't read the data type.
````

````python
FLIGHT_DF_header_schema=spark.read.format("csv")\
            .option("header", "true")\
            .option("inferschema","true")\
            .option("mode","FAILFAST")\
            .load("/FileStore/tables/2010_summary.csv")
FLIGHT_DF_header.show(5)
````

````python


FLIGHT_DF_header_schema.show(5)
````

````python
FLIGHT_DF_header_schema.printSchema() 
                                      # As inferschema is true  now it can read the data type of count. And it is showing the # actual data type of count.
````
